## 7.Solution with - Nested IF Statement

# Section C: Exploring Pandas

```python
data={
      'Name' : ['A','B','C','C','D','E','F','G','H','I'],
       'Age' : [18,19,20,20,21,22,23,24,25,26],
       'City': ['Delhi','Mumbai','Chennai','Chennai','Bangalore','Bangalore','Hyderabad','Hyderabad','Pune','Kolkatta'],
       'Salary':[10000,20000,30000,30000,40000,50000,np.nan,70000,80000,90000]
}

```

```python
df2=pd.DataFrame(data)
df2
```

```python
df2.info()
```

```python
df2.describe()
```

```python
# Select only 'City' and 'name' Columns

df2[["City","Name"]]
```

```python
df2=df2.drop(columns=["City"],errors="ignore")            #Drop the column "City"

df2["Salary"] = pd.to_numeric(df2["Salary"]).round(2)
df2["Salary"] = df2["Salary"].fillna(df2["Salary"].mean()) #Fill the "NaN" values in Salary with Salary Mean value

df2=df2.drop_duplicates()                                  # Drop duplicate records
df2
```
